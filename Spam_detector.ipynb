```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Load dataset from URL
data_url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"
print("Loading dataset...")
sms_df = pd.read_csv(data_url, sep='\t', header=None, names=['label', 'message'])

# Encode labels: 'ham' → 0, 'spam' → 1
sms_df['label'] = sms_df['label'].map({'ham': 0, 'spam': 1})

# Separate messages and their labels
texts = sms_df['message']
labels = sms_df['label']

# Split into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=42
)

# Convert text to bag-of-words vectors
vectorizer = CountVectorizer(lowercase=True, stop_words='english')
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Train a Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test_vec)

# Output evaluation metrics
print("\nModel Performance:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=["Ham", "Spam"]))

# Try the model on a new message
sample_msg = ["Congratulations! You’ve won a free vacation. Reply NOW to claim."]
sample_vec = vectorizer.transform(sample_msg)
predicted_label = model.predict(sample_vec)[0]

print("\nTest Message Prediction:")
print(f"Message: \"{sample_msg[0]}\"")
print("Prediction:", "SPAM" if predicted_label == 1 else "HAM (Not Spam)")
```
